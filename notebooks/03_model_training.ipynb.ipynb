```python
import pandas as pd
import pickle
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

df = pd.read_csv('./data/features.csv')

FEATURES = ['grid_pos', 'avg_finish_last3', 'track_temp', 'is_wet', 'driver_id']
TARGET   = 'on_podium'

X = df[FEATURES]
y = df[TARGET]

# Split: train on older seasons, test on the newest
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Random Forest is great for beginners — handles mixed data, no scaling needed
model = RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred, target_names=['No Podium', 'Podium']))

# Save the model
with open('./model/podium_model.pkl', 'wb') as f:
    pickle.dump(model, f)
```
