```python
import pandas as pd

df = pd.read_csv('./data/raw_data.csv')

# --- Target variable ---
df['on_podium'] = (df['finish_pos'] <= 3).astype(int)

# --- Feature: rolling average finishing position (last 3 races per driver) ---
df = df.sort_values(['driver', 'year', 'round'])
df['avg_finish_last3'] = (
    df.groupby('driver')['finish_pos']
      .transform(lambda x: x.shift(1).rolling(3, min_periods=1).mean())
)

# --- Feature: grid position (already collected) ---
# grid_pos is already in the dataframe

# --- Feature: encode rainfall as binary ---
df['is_wet'] = (df['rainfall'] > 0).astype(int)

# --- Feature: driver encode (label encode for simplicity) ---
df['driver_id'] = df['driver'].astype('category').cat.codes

# Drop rows with NaN (e.g. first race of a driver's career)
df = df.dropna()

df.to_csv('./data/features.csv', index=False)
print(df[['driver', 'grid_pos', 'avg_finish_last3', 'track_temp', 'is_wet', 'on_podium']].head(10))
```
