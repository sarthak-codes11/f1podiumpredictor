```python
import fastf1
import pandas as pd

fastf1.Cache.enable_cache('f1_cache')
print("FastF1 version:", fastf1.__version__)
print("✅ Everything is working!")
```

```python
records = []
seasons = [2022, 2023, 2024]

for year in seasons:
    schedule = fastf1.get_event_schedule(year, include_testing=False)
    
    for _, event in schedule.iterrows():
        round_num = event['RoundNumber']
        event_name = event['EventName']
        
        try:
            # Load race session
            race = fastf1.get_session(year, round_num, 'R')
            race.load(telemetry=False, weather=True, messages=False)
            
            # Load qualifying session
            quali = fastf1.get_session(year, round_num, 'Q')
            quali.load(telemetry=False, weather=False, messages=False)
            
            # Weather averages
            weather = race.weather_data
            avg_temp = weather['TrackTemp'].mean()
            avg_rainfall = weather['Rainfall'].mean()
            
            # Qualifying grid positions
            quali_results = quali.results[['Abbreviation', 'Position']].rename(
                columns={'Position': 'quali_pos'}
            )
            
            # Race results
            race_results = race.results[['Abbreviation', 'Position', 'GridPosition']]
            
            # Merge quali into race results
            merged = race_results.merge(quali_results, on='Abbreviation', how='left')
            
            for _, row in merged.iterrows():
                records.append({
                    'year':         year,
                    'round':        round_num,
                    'circuit':      event_name,
                    'driver':       row['Abbreviation'],
                    'grid_pos':     row['GridPosition'],
                    'quali_pos':    row.get('quali_pos', None),
                    'finish_pos':   row['Position'],
                    'track_temp':   avg_temp,
                    'rainfall':     avg_rainfall,
                })
            
            print(f"✅ {year} R{round_num} — {event_name}")
        
        except Exception as e:
            print(f"⚠️ Skipping {year} R{round_num} ({event_name}): {e}")

df = pd.DataFrame(records)

# Save to CSV
os.makedirs('data', exist_ok=True)
df.to_csv('data/raw_data.csv', index=False)

print(f"\n🏁 Done! {len(df)} rows collected across {df['round'].nunique()} rounds.")
print(df.head())
```

```python
import time
import os

records = []
seasons = [2022, 2023, 2024]  # we'll do one at a time if needed

for year in seasons:
    schedule = fastf1.get_event_schedule(year, include_testing=False)
    
    for _, event in schedule.iterrows():
        round_num = event['RoundNumber']
        event_name = event['EventName']
        
        try:
            race = fastf1.get_session(year, round_num, 'R')
            race.load(telemetry=False, weather=True, messages=False)
            
            quali = fastf1.get_session(year, round_num, 'Q')
            quali.load(telemetry=False, weather=False, messages=False)
            
            weather = race.weather_data
            avg_temp = weather['TrackTemp'].mean()
            avg_rainfall = weather['Rainfall'].mean()
            
            quali_results = quali.results[['Abbreviation', 'Position']].rename(
                columns={'Position': 'quali_pos'}
            )
            
            race_results = race.results[['Abbreviation', 'Position', 'GridPosition']]
            merged = race_results.merge(quali_results, on='Abbreviation', how='left')
            
            for _, row in merged.iterrows():
                records.append({
                    'year':       year,
                    'round':      round_num,
                    'circuit':    event_name,
                    'driver':     row['Abbreviation'],
                    'grid_pos':   row['GridPosition'],
                    'quali_pos':  row.get('quali_pos', None),
                    'finish_pos': row['Position'],
                    'track_temp': avg_temp,
                    'rainfall':   avg_rainfall,
                })
            
            print(f"✅ {year} R{round_num} — {event_name}")
            
            # ✨ KEY FIX: pause between each race to avoid rate limit
            time.sleep(3)
        
        except Exception as e:
            print(f"⚠️ Skipping {year} R{round_num} ({event_name}): {e}")
            # If rate limited, wait longer and continue
            if "500 calls/h" in str(e):
                print("⏳ Rate limit hit — waiting 60 seconds...")
                time.sleep(60)

# Save whatever we collected so far
os.makedirs('data', exist_ok=True)
df = pd.DataFrame(records)
df.to_csv('data/raw_data.csv', index=False)
print(f"\n🏁 Done! {len(df)} rows, {len(df['circuit'].unique())} races collected.")
```

```python
import pandas as pd

df = pd.read_csv('data/raw_data.csv')

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nSample rows:")
print(df.head(10))
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Drop rows with missing critical values
df = df.dropna(subset=['driver', 'grid_pos', 'quali_pos', 'finish_pos'])
print(f"Rows after cleaning: {len(df)}")

# --- Target variable ---
df['on_podium'] = (df['finish_pos'] <= 3).astype(int)
print(f"Podium rows: {df['on_podium'].sum()} / {len(df)}")

# --- Sort so rolling calculations go in the right order ---
df = df.sort_values(['driver', 'year', 'round']).reset_index(drop=True)

# --- Rolling avg finish position over last 3 races per driver ---
df['avg_finish_last3'] = (
    df.groupby('driver')['finish_pos']
      .transform(lambda x: x.shift(1).rolling(3, min_periods=1).mean())
)

# --- Rolling avg qualifying position over last 3 races ---
df['avg_quali_last3'] = (
    df.groupby('driver')['quali_pos']
      .transform(lambda x: x.shift(1).rolling(3, min_periods=1).mean())
)

# --- Wet race flag ---
df['is_wet'] = (df['rainfall'] > 0).astype(int)

# --- Encode driver as a number ---
df['driver_id'] = df['driver'].astype('category').cat.codes

# Drop rows where rolling features are NaN (very first race of a driver)
df = df.dropna(subset=['avg_finish_last3', 'avg_quali_last3'])

print(f"\nFinal dataset size: {len(df)} rows")
print(f"\nFeature preview:")
print(df[['driver', 'grid_pos', 'avg_finish_last3', 'avg_quali_last3', 'is_wet', 'on_podium']].head(10))

# Save
df.to_csv('data/features.csv', index=False)
print("\n✅ Features saved to data/features.csv")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import pickle
import os

df = pd.read_csv('data/features.csv')

FEATURES = ['grid_pos', 'quali_pos', 'avg_finish_last3', 'avg_quali_last3', 'track_temp', 'is_wet', 'driver_id']
TARGET = 'on_podium'

X = df[FEATURES]
y = df[TARGET]

# Split: last 20% of data = test set (keeps time order intact)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print(f"Training rows: {len(X_train)}, Test rows: {len(X_test)}")
print(f"Podiums in test set: {y_test.sum()} / {len(y_test)}")

# Train the model
model = RandomForestClassifier(
    n_estimators=200,
    class_weight='balanced',
    random_state=42
)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("\n📊 Classification Report:")
print(classification_report(y_test, y_pred, target_names=['No Podium', 'Podium']))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(f"  True No-Podium:  {cm[0][0]}  |  False Podium:   {cm[0][1]}")
print(f"  Missed Podiums:  {cm[1][0]}  |  Correct Podiums: {cm[1][1]}")

# Feature importance
print("\n🔍 Feature Importances:")
for feat, imp in sorted(zip(FEATURES, model.feature_importances_), key=lambda x: -x[1]):
    bar = '█' * int(imp * 50)
    print(f"  {feat:<20} {bar} {imp:.3f}")

# Save model
os.makedirs('model', exist_ok=True)
with open('model/podium_model.pkl', 'wb') as f:
    pickle.dump(model, f)

print("\n✅ Model saved to model/podium_model.pkl")
```

```python
import pickle
import pandas as pd

# Load model
with open('model/podium_model.pkl', 'rb') as f:
    model = pickle.load(f)

# Load driver encoding from training data
df = pd.read_csv('data/features.csv')
driver_encoding = df[['driver', 'driver_id']].drop_duplicates().set_index('driver')['driver_id'].to_dict()

# --- Fill in qualifying results for a race you want to predict ---
# Example: 2024 British GP qualifying order
upcoming = pd.DataFrame([
    {'driver': 'VER', 'grid_pos': 1, 'quali_pos': 1},
    {'driver': 'NOR', 'grid_pos': 2, 'quali_pos': 2},
    {'driver': 'SAI', 'grid_pos': 3, 'quali_pos': 3},
    {'driver': 'HAM', 'grid_pos': 4, 'quali_pos': 4},
    {'driver': 'RUS', 'grid_pos': 5, 'quali_pos': 5},
    {'driver': 'LEC', 'grid_pos': 6, 'quali_pos': 6},
    {'driver': 'PIA', 'grid_pos': 7, 'quali_pos': 7},
    {'driver': 'ALO', 'grid_pos': 8, 'quali_pos': 8},
    {'driver': 'PER', 'grid_pos': 9, 'quali_pos': 9},
    {'driver': 'STR', 'grid_pos': 10, 'quali_pos': 10},
])

# Fill in race conditions + recent form from your dataset
# Using 2024 averages as a rough estimate
for col in ['avg_finish_last3', 'avg_quali_last3']:
    driver_avgs = df.groupby('driver')[col].mean()
    upcoming[col] = upcoming['driver'].map(driver_avgs).fillna(10.0)

upcoming['track_temp'] = 22   # British GP is usually cool
upcoming['is_wet'] = 0
upcoming['driver_id'] = upcoming['driver'].map(driver_encoding).fillna(-1)

# Predict
FEATURES = ['grid_pos', 'quali_pos', 'avg_finish_last3', 'avg_quali_last3', 'track_temp', 'is_wet', 'driver_id']
probs = model.predict_proba(upcoming[FEATURES])[:, 1]
upcoming['podium_probability'] = (probs * 100).round(1)

# Show results
result = upcoming[['driver', 'grid_pos', 'podium_probability']].sort_values('podium_probability', ascending=False)
result['predicted_podium'] = ['🏆' if i < 3 else '' for i in range(len(result))]
print(result.to_string(index=False))
```
